```python
import numpy as np

a=np.array([[[10,20],[30,40]]])
b=np.array([[[50,60],[70,80]]])

print(np.sum(a))
print(np.sum(b))
print(np.min(a))
print(np.min(b))
print(np.max(a))
print(np.max(b))
print(np.mean(a))
print(np.std(a))
```

```python
import numpy as np

a=np.array([[[10,20],[30,40]]])
b=np.array([[[50,60],[70,80]]])

print(a[(a > 20) & (a < 40)])

print(b[(b > 50) & (b < 80)])

print(np.sum(a,axis=0))
print(np.sum(a,axis=1))
```
